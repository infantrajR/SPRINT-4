```python
import pandas as pd

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

display(df.head())
```

```python
print("First 5 rows:")
display(df.head())

print("Last 5 rows:")
display(df.tail())

print("Dataset information:")
df.info()

print("Statistical summary:")
display(df.describe(include="all"))
```

```python
# Identify numerical and categorical features

numerical_features = df.select_dtypes(include="number").columns.tolist()
categorical_features = df.select_dtypes(exclude="number").columns.tolist()

print("Numerical features:")
print(numerical_features)

print("\nCategorical features:")
print(categorical_features)

print("\nTotal numerical features:", len(numerical_features))
print("Total categorical features:", len(categorical_features))
```

```python
print("Target variable: Survived")

print("\nSurvival counts:")
print(df["Survived"].value_counts())

print("\nSurvival percentages:")
print((df["Survived"].value_counts(normalize=True) * 100).round(2))
```

```python
print("Missing values in each column:")
display(df.isnull().sum())

print("\nTotal missing values:")
print(df.isnull().sum().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
import matplotlib.pyplot as plt

df["Survived"].value_counts().sort_index().plot(
    kind="bar",
    title="Titanic Passenger Survival Distribution"
)

plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")
plt.xticks(rotation=0)
plt.show()
```

# Dataset Understanding — Findings

The Titanic dataset contains passenger information, including age, sex, passenger class, fare, and survival status. The dataset contains numerical and categorical features. The target variable is Survived, where 0 indicates that a passenger did not survive and 1 indicates that a passenger survived. Missing values and duplicate records were examined to assess data quality. A bar chart was created to visualize the distribution of passenger survival. This initial analysis helps identify the dataset structure and potential data-quality issues before further analysis.
