# Categorical Data Analysis

Sample Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Sample customer records; Purchased is the target variable.
df = pd.DataFrame({
    "Customer_ID": range(101, 117),
    "Department": ["IT", "HR", "Sales", "IT", "Finance", "IT", "HR", "Sales", "IT", "Finance", "IT", "Sales", "HR", "IT", "Legal", "IT"],
    "Membership": ["Gold", "Silver", "Bronze", "Gold", "Silver", "Gold", "Bronze", "Silver", "Gold", "Bronze", "Silver", "Gold", "Silver", "Gold", "Bronze", "Gold"],
    "City": ["Bengaluru", "Mysuru", "Hassan", "Bengaluru", "Delhi", "Mumbai", "Mysuru", "Pune", "Chennai", "Hassan", "Delhi", "Kochi", "Mumbai", "Pune", "Jaipur", "Bengaluru"],
    "Purchased": ["Yes", "No", "No", "Yes", "No", "Yes", "No", "Yes", "Yes", "No", "No", "Yes", "No", "Yes", "No", "Yes"]
})
print(df.head())
```

# 1. Unique Categories
# Definition
Unique categories are the distinct labels found in a categorical column. Repeated labels are counted only once. Checking unique categories helps us understand which groups are present.

# Business Example
A company checks which departments appear in employee records. This can reveal a new department or an unexpected spelling. The company can then standardize category labels.

# AI/ML Use
Unique labels show which categories an ML model may need to encode. They also reveal unexpected labels before preprocessing. Clean categories help avoid inconsistent predictions.

# Python Example

```python
print("Unique departments:", df["Department"].unique())
```

# Explanation
The unique() method lists each department once. This helps us inspect the available labels.

# 2. Category Frequencies
# Definition
Category frequency is the number of times each label appears. It tells us how common each category is. Frequency counts make comparisons between groups simple.

# Business Example
A retailer counts customers in each membership level. The counts help estimate demand from different customer groups. Staff can use the result for planning.

# AI/ML Use
Frequency counts identify dominant and uncommon classes. This helps select encoding methods and examine sampling needs. It can also expose poorly represented categories.

# Python Example

```python
print(df["Membership"].value_counts())
```

# Explanation
value_counts() shows how many customers belong to each membership group.

# 3. Category Distribution
# Definition
Category distribution describes how records are divided among categories. Percentages make groups easy to compare even across datasets of different sizes. A distribution may be balanced or concentrated in a few groups.

# Business Example
A business studies the percentage of customers in each department. This helps understand its customer mix. Percentage shares can be compared over time.

# AI/ML Use
Category percentages help evaluate whether training data represents the expected population. Large differences may signal sampling bias. The result supports fair evaluation and preprocessing.

# Python Example

```python
print((df["Department"].value_counts(normalize=True) * 100).round(1))
```

# Explanation
normalize=True converts counts into proportions. Multiplying by 100 gives percentages.

# 4. Rare Categories
# Definition
Rare categories are labels that appear very few times. They can represent genuine small groups or mistakes in the data. Their importance depends on the business context.

# Business Example
A company finds only one customer from a particular department. That department might be legitimate rather than an error. The team checks its meaning before combining or removing it.

# AI/ML Use
Rare categories may create unstable patterns in a model. Some encoding methods produce sparse features for them. We should check whether rare labels are meaningful before grouping them.

# Python Example

```python
counts = df["Department"].value_counts()
print("Categories occurring fewer than 2 times:")
print(counts[counts < 2])
```

# Explanation
This code flags categories with fewer than two records. A flagged category is not automatically invalid.

# 5. High Cardinality
# Definition
High cardinality means a categorical feature contains many different labels. It often occurs in city names, product IDs, or user IDs. A large number of categories can make analysis and encoding harder.

# Business Example
An online store delivers to many cities. The City column may contain dozens or thousands of distinct locations. The business may group locations by region when appropriate.

# AI/ML Use
One-hot encoding a high-cardinality feature can create many columns. This can increase memory use and overfitting risk. We must choose encoding based on the feature and model.

# Python Example

```python
print("Unique cities:", df["City"].nunique())
print("Unique customer IDs:", df["Customer_ID"].nunique())
```

# Explanation
nunique() counts distinct values. Customer_ID has a unique value for each record and should usually be treated as an identifier, not a predictive category.

# 6. Low Cardinality
# Definition
Low cardinality means a categorical column has only a few distinct labels. Examples include Yes/No and membership levels. These variables are usually easy to summarize and encode.

# Business Example
A shop records whether each customer purchased a product. There are only two labels: Yes and No. The business can compare purchase rates directly.

# AI/ML Use
Low-cardinality variables are often suitable for one-hot or binary encoding. They can provide useful model signals. We still need to check class balance and missing labels.

# Python Example

```python
print("Purchased categories:", df["Purchased"].unique())
print("Number of categories:", df["Purchased"].nunique())
```

# Explanation
The target Purchased contains two distinct labels, making it a binary target.

# 7. Category Imbalance
# Definition
Category imbalance occurs when some labels appear much more often than others. It is especially important for a classification target. A high overall accuracy can hide poor performance on the minority class.

# Business Example
A bank may have many normal transactions but few fraudulent transactions. Predicting every transaction as normal could appear accurate. The bank must measure fraud detection separately.

# AI/ML Use
Imbalanced target classes require careful evaluation. Precision, recall, F1-score, and stratified splitting can be useful. We should inspect the distribution before choosing a strategy.

# Python Example


```python
print("Purchase counts:")
print(df["Purchased"].value_counts())
print("Purchase percentages:")
print((df["Purchased"].value_counts(normalize=True) * 100).round(1))
```

# Explanation
The code checks the number and share of Yes and No outcomes. Whether imbalance is serious depends on the use case and dataset size.

# 8. Categorical Variables vs Target Variable
# Definition
This analysis compares a categorical input feature with a target variable. It shows how the target outcome changes across different groups. We can use cross-tabulations and within-group percentages to compare relationships.

# Business Example
A retailer compares purchase decisions across membership levels. The results show which groups purchase more often in the sample. The business can use this as an initial finding for further investigation.

# AI/ML Use
Comparing features with the target helps identify potentially predictive categories. It may reveal imbalance within groups and possible leakage concerns. A relationship in a small sample does not prove causation or guarantee predictive power.

# Python Example

```python
# Counts of purchases by membership level
print(pd.crosstab(df["Membership"], df["Purchased"]))

# Purchase rate within each membership level
rates = pd.crosstab(df["Membership"], df["Purchased"], normalize="index") * 100
print("\nPurchase percentages by membership:\n", rates.round(1))

# Compare a second categorical feature with the target
print("\nDepartment vs Purchased:\n", pd.crosstab(df["Department"], df["Purchased"]))
```

# Explanation
pd.crosstab() compares two categorical columns. The first table gives counts; normalize="index" gives percentages within each membership group. Compare the Yes percentages to see which groups purchase more often. These are descriptive findings from sample data, not proof of a causal relationship.
